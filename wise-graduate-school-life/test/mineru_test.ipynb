```python
from pathlib import Path

# asset 폴더의 PDF 파일 목록
pdf_files = sorted(Path("../asset").glob("*.pdf"))
for f in pdf_files:
    print(f"  - {f.name}")

# 테스트할 PDF 선택 (첫 번째 파일)
pdf_path = str(pdf_files[0])
print(f"\n선택된 PDF: {pdf_path}")
```

```python
import subprocess

result = subprocess.run(
    [
        "../.venv/bin/mineru",
        "-p",
        pdf_path,
        "-o",
        "output/mineru_result",
        "-b",
        "pipeline",  # CPU 가능, 가벼움
        "-l",
        "en",
    ],  # 영어 논문
    capture_output=True,
    text=True,
)
```
