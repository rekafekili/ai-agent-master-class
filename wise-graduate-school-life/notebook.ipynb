```python
from dotenv import load_dotenv
load_dotenv()

from langgraph.types import Command
from agent.main_agent import graph
```

```python
config = {"configurable": {"thread_id": "paper-session-01"}}

# 1단계: 파싱 → 단어 추출 → human_review에서 interrupt (사용자 입력 대기)
result = graph.invoke(
    {
        "file_path": "asset/MSCRS_ Multi-modal Semantic Graph Prompt Learning Framework for Conversational Recommender Systems.pdf",
    },
    config=config,
)

# interrupt 상태 확인
state = graph.get_state(config)
print(f"논문: {state.values['paper_title']}")
print(f"섹션 수: {len(state.values['sections'])}")
print(f"추출된 단어 수: {len(state.values['vocabulary'])}")
print()
for v in state.values["vocabulary"]:
    print(f"  [{v['section_index']}] {v['word']} — {v['meaning_ko']}")
print()
print("--- 다음 셀에서 저장 여부를 결정하세요 ---")
```

```python
# 2단계: interrupt에 응답하여 resume
# save=True → save_vocabulary 노드 → END
# save=False → 바로 END
result = graph.invoke(
    Command(resume={"save": True}),
    config=config,
)
print("완료!")
```
