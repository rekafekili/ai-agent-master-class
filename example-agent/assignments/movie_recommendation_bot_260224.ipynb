```python
# Q1. 지금 인기 있는 영화가 무엇인지 알려줘.
# Q2. movie ID 550에 해당하는 영화가 무엇인지 알려줘
# Q3. movie ID 550에 해당하는 영화에 누가 출연하는지 알려줘.
import requests

base_url = "https://nomad-movies.nomadcoders.workers.dev"

def get_popular_movies():
    res = requests.get(f"{base_url}/movies")
    return res.content

def get_movie_details(id):
    res = requests.get(f"{base_url}/movies/{id}")
    return res.content

def get_movie_credits(id):
    res = requests.get(f"{base_url}/movies/{id}/credits")
    return res.content
```

```python
from dotenv import load_dotenv
import os
import openai

load_dotenv()
client = openai.OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

PROMPT = """
I have the following functions.

'get_popular_movies()' : To get popular movie list.
'get_movie_details(id)' : To get movie info by movie id
'get_movie_credits(id)' : To get movie credits by movie id

Please answer with the name of the function that you would like me to run.

Please say nothing else, just the name of the function with the arguments.
"""

messages = []
```

```python
def call_ai():
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages
    )
    aiMessage = response.choices[0].message.content
    print(f"AI: {aiMessage}")
    messages.append({
        "role": "assistant",
        "content": aiMessage
    })
```

```python
while True:
    userMessage = input("Enter a user message...")
    if userMessage == 'quit' or userMessage == 'q':
        break
    else:
        print(f"User: {userMessage}")
        messages.append({
            "role": "user",
            "content": userMessage
        })
        call_ai()
        
```
