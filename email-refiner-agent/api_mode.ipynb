````python
import requests

BASE_URL = "http://127.0.0.1:8000"
APP_NAME = "travel_advisor_agent"
USER_ID = "u_123"
SESSION_ID = "8700e916-bb53-4b0c-8112-944c293f49c5"
````

````python
# Create Session (/apps/{app_name}/users/{user_id}/sessions)
response = requests.post(f"{BASE_URL}/apps/{APP_NAME}/users/{USER_ID}/sessions")

print(response.json())
````

````python
import sseclient
import json

message = {
    "appName": APP_NAME,
    "userId": USER_ID,
    "sessionId": SESSION_ID,
    "newMessage": {
        "parts": [
            {"text": "I want to go Berlin",}
        ],
        "role": "user"
    },
    "streaming": True,
}

response = requests.post(
    f"{BASE_URL}/run_sse",
    json=message,
    stream=True,
)

client = sseclient.SSEClient(response)

for event in client.events():
    data = json.loads(event.data)

    content = data.get("content")
    parts = content.get("parts")
    for part in parts:
        function_call = part.get('fuctionCall', None)
        if function_call:
            print(function_call.get("name"))
        text = part.get("text", None)
        if text:
            print(text)
    print("+"*60)

# print(response.json())
````
